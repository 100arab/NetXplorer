```python
# Experiment-2
# 23015063
# Saurabh Jawade
# Perform data cleaning, handling missing values, normalization,
# encoding categorical features using a network security dataset.

import pandas as pd
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# Load dataset
df = pd.read_csv("network_security_dataset.csv")

print("Original Dataset:")
print(df)

# Remove duplicate rows
df = df.drop_duplicates()

# Check missing values
print("\nMissing values before cleaning:")
print(df.isnull().sum())

# Fill missing numeric values with mean
numeric_cols = ["src_bytes", "dst_bytes", "duration", "packet_count", "login_attempts"]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Fill missing categorical values with mode
categorical_cols = ["protocol", "service", "flag"]

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Encode categorical features
label_encoder = LabelEncoder()

for col in categorical_cols:
    df[col] = label_encoder.fit_transform(df[col])

# Encode target label
df["label"] = label_encoder.fit_transform(df["label"])

# Normalize numeric features
scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("\nCleaned Dataset:")
print(df)

# Save cleaned dataset
df.to_csv("cleaned_network_security_dataset.csv", index=False)

print("\nCleaned CSV file saved successfully.")
```
